```python
import pandas as pd # Manipular los datos
import matplotlib.pyplot as plt ## Poder hacer las graficas
import matplotlib.ticker as mticker 
df = pd.read_csv('/home/santiago/Documentos/archivosPruebas/avocado.csv', index_col=0)
df.head(10)

fig, ax = plt.subplots(figsize=(8,5))
ax.hist(df['AveragePrice'], bins=30, color='mediumseagreen', edgecolor='White')
ax.set_title('Distribución de distintos precios de los aguacates')
ax.set_xlabel('precios en USD')
ax.set_ylabel('Frecuencia de ventas')
plt.tight_layout()
plt.show()
```
